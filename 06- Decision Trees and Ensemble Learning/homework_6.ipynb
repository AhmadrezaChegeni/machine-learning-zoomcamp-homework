```python
import numpy as np
import pandas as pd
import xgboost as xgb
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
from sklearn.tree import DecisionTreeRegressor, export_text
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
```

Preparing the dataset

```python
data = 'https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv'
```

```python
df = pd.read_csv(data)
```

```python
df= df.fillna(0)
```

```python
df.isnull().sum()
```

```python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1)

df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

y_train = df_train.fuel_efficiency_mpg.values
y_val = df_val.fuel_efficiency_mpg.values
y_test = df_test.fuel_efficiency_mpg.values
```

```python
del df_train['fuel_efficiency_mpg']
del df_val['fuel_efficiency_mpg']
del df_test['fuel_efficiency_mpg']
```

```python
dv = DictVectorizer(sparse=True)
X_train = dv.fit_transform(df_train.to_dict(orient='records'))
X_val = dv.transform(df_val.to_dict(orient='records'))
```

Question 2

```python
rf = RandomForestRegressor(n_estimators=10, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)
```

```python
from sklearn.metrics import root_mean_squared_error
y_pred =  rf.predict(X_val)
rmse = root_mean_squared_error(y_val, y_pred)

print('Validation RMSE:', rmse)
```

Question 3

```python
scores = []

for n in [10, 50, 100, 150]:

    rf = RandomForestRegressor(
        n_estimators=n,
        random_state=1,
        n_jobs=-1
    )

    rf.fit(X_train, y_train)

    y_pred = rf.predict(X_val)

    rmse = root_mean_squared_error(y_val, y_pred)

    scores.append((n, rmse))

    print(f'n_estimators={n}, RMSE={rmse:.4f}') 
```

Question 4

```python
scores = []

for depth in [10, 15, 20, 25]:

    for n in [10, 50, 100, 150]:

        rf = RandomForestRegressor(
            n_estimators=n,
            max_depth=depth,
            random_state=1,
            n_jobs=-1
        )

        rf.fit(X_train, y_train)

        y_pred = rf.predict(X_val)

        rmse = root_mean_squared_error(y_val, y_pred)

        scores.append({
            'max_depth': depth,
            'n_estimators': n,
            'rmse': rmse
        })

        print(
            f'depth={depth}, '
            f'n_estimators={n}, '
            f'RMSE={rmse:.4f}'
        )
```

```python
results = pd.DataFrame(scores)

print(results)
```

```python
mean_scores = (
    results
    .groupby('max_depth')['rmse']
    .mean()
    .reset_index()
)

print(mean_scores)
```

```python
best_depth = mean_scores.loc[
    mean_scores['rmse'].idxmin()
]

print('Best max_depth:', best_depth['max_depth'])
print('Mean RMSE:', best_depth['rmse'])
```

```python
rf = RandomForestRegressor(
    n_estimators=10,
    max_depth=20,
    random_state=1,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
importances = rf.feature_importances_

feature_names = dv.get_feature_names_out()

feature_importance_df = pd.DataFrame({
    'feature': feature_names,
    'importance': importances
})
```

```python
feature_importance_df = feature_importance_df.sort_values(
    by='importance',
    ascending=False
)

print(feature_importance_df.head(10))
```

```python

candidates = [
    'vehicle_weight',
    'horsepower',
    'acceleration',
    'engine_displacement'
]

candidate_importances = feature_importance_df[
    feature_importance_df['feature'].isin(candidates)
]

print(candidate_importances)
```

Question 6

```python
features = list(dv.get_feature_names_out())

dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=features)
dval = xgb.DMatrix(X_val, label=y_val, feature_names=features)

watchlist = [(dtrain, 'train'), (dval, 'val')]
```

```python
for eta in [0.3, 0.1]:

    xgb_params = {
        'eta': eta,
        'max_depth': 6,
        'min_child_weight': 1,
        'objective': 'reg:squarederror',
        'nthread': 8,
        'seed': 1,
        'verbosity': 1
    }

    model = xgb.train(
        xgb_params,
        dtrain,
        num_boost_round=100,
        evals=watchlist,
        verbose_eval=False
    )

    y_pred = model.predict(dval)

    rmse = root_mean_squared_error(y_val, y_pred)

    print(f'eta = {eta}, RMSE = {rmse:.5f}')
```
